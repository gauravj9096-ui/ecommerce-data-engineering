```python
txn=spark.read.csv(RAW_PATH+'transactions.csv',header=True,inferSchema=True)
cust =spark.read.csv(RAW_PATH+'customers.csv',header=True,inferSchema=True)
prod =spark.read.csv(RAW_PATH+'products.csv',header=True,inferSchema=True)
store =spark.read.csv(RAW_PATH+'stores.csv',header=True,inferSchema=True)
promo =spark.read.csv(RAW_PATH+'promotions.csv',header=True,inferSchema=True)
fb=spark.read.csv(RAW_PATH+'feedbacks.csv',header=True,inferSchema=True)
```

```python
txn.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'transactions')
cust.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'customers')
prod.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'products')
store.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'stores')
promo.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'promotions')
fb.write.format('Delta').mode('overwrite').save(BRONZE_PATH+'feedbacks')
```
